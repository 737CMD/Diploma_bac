```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import re
```

```python
def load(path):
    with open(path, 'r') as f:
        data = re.findall(r'([-\d\.eE]+)[\s,]+([-\d\.eE]+)', f.read())
    return pd.DataFrame(data, columns=['x', 'cp'], dtype=float)

try:
    u, l, exp = load('upper7.txt'), load('lower7.txt'), load('cp(x)7-pres.txt')
    
    plt.figure(figsize=(10, 6))
    plt.plot(u.x, u.cp, 'bo', mfc='none', ms=4)
    plt.plot(l.x, l.cp, 'bo', mfc='none', ms=4, label='AGARD')
    plt.plot(exp.x, exp.cp, 'ko', ms =3, lw=1.5, label='Pressure based RANS-SA')
    plt.title("Cp(x) Rae2822 AGARD 138 case 7")
    plt.gca().invert_yaxis()
    plt.grid(True, ls=':', alpha=0.6)
    plt.xlabel('x/c'); plt.ylabel('Cp')
    plt.legend()
    plt.tight_layout()
    plt.savefig('cp6_plot.png', dpi=300)
    plt.show()

except Exception as e:
    print(f"Error: {e}")
```

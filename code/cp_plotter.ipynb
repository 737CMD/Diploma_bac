```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import re
```

```python
def load(path):
    with open(path, 'r') as f:
        data = re.findall(r'([-\d\.eE]+)[\s,]+([-\d\.eE]+)', f.read())
    return pd.DataFrame(data, columns=['x', 'cp'], dtype=float)

try:
    u, l, exp = load('upper8.txt'), load('lower8.txt'), load('cp(x)8.txt')
    
    plt.figure(figsize=(10, 6))
    plt.plot(u.x, u.cp, 'bo', mfc='none', ms=4, label='Digitized Upper')
    plt.plot(l.x, l.cp, 'bo', mfc='none', ms=4, label='Digitized Lower')
    plt.plot(exp.x, exp.cp, 'k--', lw=1.5, label='Numerical Exp')

    plt.gca().invert_yaxis()
    plt.grid(True, ls=':', alpha=0.6)
    plt.xlabel('x/c'); plt.ylabel('Cp')
    plt.legend()
    plt.tight_layout()
    plt.savefig('cp_plot.png', dpi=300)
    plt.show()

except Exception as e:
    print(f"Error: {e}")
```

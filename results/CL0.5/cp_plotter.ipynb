```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import re
```

```python
def load(path):
    with open(path, 'r') as f:
        data = re.findall(r'([-\d\.eE]+)[\s,]+([-\d\.eE]+)', f.read())
    return pd.DataFrame(data, columns=['x', 'cp'], dtype=float)

try:
    rae, exp1, exp2 = load('rae2822cp.txt'), load('ego99.txt'), load('ego112.txt')
    
    plt.figure(figsize=(10, 6))
    plt.plot(rae.x, rae.cp, 'bo', mfc='none', ms=4, label=r'RAE2822  $C_{xa}$ = 0.0107')
    plt.plot(exp1.x, exp1.cp, 'ro', mfc='none', ms=4, label=r'EGO $C_{xa}$ = 0.0099')
    plt.plot(exp2.x, exp2.cp, 'ko', ms =3, lw=1.5, label=r'EGO $C_{xa}$ = 0.0112')
    plt.title(r"Распределение $C_p$(x) для RAE2822 и результатов работы EGO")
    plt.gca().invert_yaxis()
    plt.grid(True, ls=':', alpha=0.6)
    plt.xlabel('x/c'); plt.ylabel(r'$C_p$')
    plt.legend()
    plt.tight_layout()
    plt.savefig('cp6_plot.png', dpi=300)
    plt.show()

except Exception as e:
    print(f"Error: {e}")
```
